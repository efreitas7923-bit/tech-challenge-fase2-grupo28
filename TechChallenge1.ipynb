```python
# Carrega o arquivo application_record.csv novamente em um novo DataFrame
df_application = pd.read_csv(file_path)

# Define o dicionário de tradução para as colunas do application_record.csv
app_col_translation = {
    'ID': 'ID',
    'CODE_GENDER': 'GENERO',
    'FLAG_OWN_CAR': 'POSSUI_CARRO',
    'FLAG_OWN_REALTY': 'POSSUI_IMOVEL',
    'CNT_CHILDREN': 'NUM_FILHOS',
    'AMT_INCOME_TOTAL': 'RENDA_TOTAL',
    'NAME_INCOME_TYPE': 'TIPO_RENDA',
    'NAME_EDUCATION_TYPE': 'NIVEL_EDUCACAO',
    'NAME_FAMILY_STATUS': 'STATUS_CIVIL',
    'NAME_HOUSING_TYPE': 'TIPO_MORADIA',
    'DAYS_BIRTH': 'DIAS_NASCIMENTO',
    'DAYS_EMPLOYED': 'DIAS_EMPREGADO',
    'FLAG_MOBIL': 'FLAG_CELULAR',
    'FLAG_WORK_PHONE': 'FLAG_TELEFONE_TRABALHO',
    'FLAG_PHONE': 'FLAG_TELEFONE',
    'FLAG_EMAIL': 'FLAG_EMAIL',
    'OCCUPATION_TYPE': 'TIPO_OCUPACAO',
    'CNT_FAM_MEMBERS': 'NUM_MEMBROS_FAMILIA'
}

# Renomeia as colunas do df_application
df_application = df_application.rename(columns=app_col_translation)

# Exibe as primeiras linhas do DataFrame com as colunas traduzidas
display(df_application.head())
```

Vamos garantir que a coluna `NUM_MEMBROS_FAMILIA` seja do tipo `int64`.

```python
# Converte a coluna 'NUM_MEMBROS_FAMILIA' de float64 para int64
df_application['NUM_MEMBROS_FAMILIA'] = df_application['NUM_MEMBROS_FAMILIA'].astype('int64')

# Exibe as informações atualizadas do DataFrame para confirmar a mudança
df_application.info()
```

Vamos converter a coluna `DIAS_NASCIMENTO` para uma nova coluna `IDADE`, que representará a idade em anos.

```python
# Converte a coluna 'DIAS_NASCIMENTO' (que está em dias negativos) para 'IDADE' em anos.
# Dividimos por 365.25 para uma estimativa mais precisa considerando anos bissextos e convertemos para int.
df_application['IDADE'] = (-df_application['DIAS_NASCIMENTO'] / 365.25).astype('int64')

# Exibe as colunas 'DIAS_NASCIMENTO' e a nova coluna 'IDADE' para verificação
print('Colunas DIAS_NASCIMENTO e IDADE após a conversão:')
display(df_application[['DIAS_NASCIMENTO', 'IDADE']].head())
```

Vamos converter a coluna `DIAS_EMPREGADO` em anos de emprego, criando uma nova coluna `ANOS_EMPREGADO`. Consideraremos também a condição da coluna `TIPO_OCUPACAO` e o valor especial de `DIAS_EMPREGADO` (365243).

```python
import numpy as np

# Cria uma nova coluna 'ANOS_EMPREGADO' inicializada com NaN (Not a Number)
df_application['ANOS_EMPREGADO'] = np.nan

# Preenche 'ANOS_EMPREGADO' para casos em que DIAS_EMPREGADO é 365243 (geralmente significa desempregado)
df_application.loc[df_application['DIAS_EMPREGADO'] == 365243, 'ANOS_EMPREGADO'] = 0

# Preenche 'ANOS_EMPREGADO' para os dias de emprego normais (valores negativos)
# Apenas para as linhas onde DIAS_EMPREGADO é negativo e ANOS_EMPREGADO ainda não foi definido (não é 0)
mask_empregado = (df_application['DIAS_EMPREGADO'] < 0) & (df_application['ANOS_EMPREGADO'].isnull())
df_application.loc[mask_empregado, 'ANOS_EMPREGADO'] = ((-df_application.loc[mask_empregado, 'DIAS_EMPREGADO']) / 365.25).astype(int)

# Aplica a condição: se TIPO_OCUPACAO estiver vazio, defina ANOS_EMPREGADO como "Não informado"
df_application.loc[df_application['TIPO_OCUPACAO'].isnull(), 'ANOS_EMPREGADO'] = "Não informado"

# Exibe as colunas relevantes para verificação
print('Colunas DIAS_EMPREGADO, TIPO_OCUPACAO e ANOS_EMPREGADO após as transformações:')
display(df_application[['DIAS_EMPREGADO', 'TIPO_OCUPACAO', 'ANOS_EMPREGADO']].head(10))
```
